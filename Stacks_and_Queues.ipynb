stack using custom class

```python
class Stack:
    def __init__(self):
        self.items = []

    def push(self, item):
        self.items.append(item)

    def pop(self):
        if not self.is_empty():
            return self.items.pop()
        return "Stack is empty"

    def peek(self):
        if not self.is_empty():
            return self.items[-1]
        return "Stack is empty"

    def is_empty(self):
        return len(self.items) == 0

    def size(self):
        return len(self.items)


# Example usage
stack = Stack()
stack.push(10)
stack.push(20)
stack.push(30)
print("Top element:", stack.peek())
print("Popped element:", stack.pop())
print("Stack size:", stack.size())
```

stack using queues

```python
from collections import deque
class MyStack:

    def __init__(self):
        #initialize two queues
        self.q1 = deque()
        self.q2 = deque()


    def push(self, x: int) -> None:
        #enque x into q2
        self.q2.append(x)

        #move all elements from q1 to q2
        while self.q1:
            self.q2.append(self.q1.popleft())

        #swap q1 and q2
        self.q1, self.q2 = self.q2, self.q1



    def pop(self) -> int:
        return self.q1.popleft()


    def top(self) -> int:
        return self.q1[0]


    def empty(self) -> bool:
        return len(self.q1) == 0

# Your MyStack object will be instantiated and called as such:
# obj = MyStack()
# obj.push(x)
# param_2 = obj.pop()
# param_3 = obj.top()
# param_4 = obj.empty()
```
