```python
class Node:
  def __init__ (self, data):
    self.data = data
    self.next = None
```

```python
head = Node(10)
```

```python
head.next = Node(20)
head.next.next = Node(30)
head.next.next.next = Node(40)
head.next.next.next.next = Node(50)
```

```python
temp = head
while temp is not None:
  print (temp.data , end = " ")
  temp = temp.next
```
